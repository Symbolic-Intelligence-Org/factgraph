# Full Round Story Demo

This notebook is the readable presentation layer for the canonical demo script. The Python script remains the behavioral source of truth: `examples/round_story_full_demo.py`.

The demo uses the default native engine and covers SDK-authored setup, Q1-Q5 application capabilities, ProofFrame recheck, the three rule overlay operations, durable round events, ProofFrame diff, and the v0.1 public-boundary note.

```python
import sys
from importlib import util
from pathlib import Path

repo_root = Path.cwd()
if not (repo_root / "examples" / "round_story_full_demo.py").exists():
    repo_root = repo_root.parent

demo_path = repo_root / "examples" / "round_story_full_demo.py"
spec = util.spec_from_file_location("round_story_full_demo", demo_path)
assert spec is not None and spec.loader is not None
module = util.module_from_spec(spec)
sys.modules[spec.name] = module
spec.loader.exec_module(module)

outcomes = module.run_demo(verbose=True)
assert outcomes == module.EXPECTED_PHASE_SUMMARY
outcomes
```
